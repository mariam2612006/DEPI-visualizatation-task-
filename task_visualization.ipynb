```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
arsenal=df[df["Team"]=="Arsenal"]
sns.lineplot(data=arsenal,x="Season", y="Points")
plt.title("Arsenal Points Across All Seasons")
plt.xlabel("season")
plt.ylabel("points")
```

```python
#47
team_data = df[
    (df["Team"] == "Arsenal") &
    (df["Season"] == "2023-24")
]

results = pd.DataFrame({
    "Result": ["Wins", "Draws", "Losses"],
    "Count": [
        team_data["Wins"].sum(),
        team_data["Draws"].sum(),
        team_data["Losses"].sum()
    ]
})

px.pie(
    results,
    values="Count",
    names="Result",
    title="Arsenal Wins, Draws, and Losses in 2023-24"
)
```

```python
#48
team_data = df[
    (df["Team"] == "Liverpool") &
    (df["Season"] == "2024-25")
]

results = pd.DataFrame({
    "Result": ["Wins", "Draws", "Losses"],
    "Count": [
        team_data["Wins"].sum(),
        team_data["Draws"].sum(),
        team_data["Losses"].sum()
    ]
})

px.pie(
    results,
    values="Count",
    names="Result",
    title="Liverpool Wins, Draws, and Losses in 2024-25"
)
```

```python
#49
season_data = df[df["Season"] == "2023-24"]

performance_count = season_data["Performance_Level"].value_counts().reset_index()

performance_count.columns = ["Performance_Level", "Count"]

px.pie(
    performance_count,
    values="Count",
    names="Performance_Level",
    title="Teams in Each Performance Level in 2023-24"
)
```
